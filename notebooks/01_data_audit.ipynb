```python
from pathlib import Path
import pandas as pd

# Get the current notebook directory
NOTEBOOK_DIR = Path.cwd()

# Locate the CSV beside the notebook
FILE_PATH = NOTEBOOK_DIR / "bank-additional-full.csv"

# Verify the file exists
print("Notebook directory:", NOTEBOOK_DIR)
print("CSV file exists:", FILE_PATH.exists())
```

```python
df_raw = pd.read_csv(FILE_PATH, sep=";")

print("Dataset loaded successfully!")
print("Rows:", df_raw.shape[0])
print("Columns:", df_raw.shape[1])
```

```python
# Inspect column names, data types, and non-null counts
df_raw.info()
```

```python
# Display the first five records
display(df_raw.head())
```

```python
# Review summary statistics for all columns
display(df_raw.describe(include="all").T)
```

```python
# 1. Check conventional missing values
print("Total missing cells:", df_raw.isna().sum().sum())

# 2. Check exact duplicate rows
print("Exact duplicate rows:", df_raw.duplicated().sum())

# 3. Inspect categorical values
categorical_cols = df_raw.select_dtypes(
    include=["object", "str"]
).columns

for col in categorical_cols:
    print(f"\n{col}:")
    print(df_raw[col].value_counts(dropna=False))
```

```python
# Create a working copy
df_clean = df_raw.copy()

# Remove exact duplicate rows from the working copy
rows_before = len(df_clean)

df_clean = df_clean.drop_duplicates().copy()

rows_after = len(df_clean)

print("Original records:", rows_before)
print("Prepared records:", rows_after)
print("Duplicate occurrences removed:", rows_before - rows_after)

# Verify the cleaning result
print("Remaining duplicate rows:", df_clean.duplicated().sum())
print("Remaining null cells:", df_clean.isna().sum().sum())
```

```python
# Create the processed-data directory beside the notebook
PROCESSED_DIR = Path("data") / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

# Save the prepared dataset
output_path = PROCESSED_DIR / "bank_marketing_prepared.csv"

df_clean.to_csv(output_path, index=False)

print("Prepared dataset saved to:", output_path.resolve())
print("File exists:", output_path.exists())

```
